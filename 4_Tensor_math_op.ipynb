```python
import torch
```

```python
tensor_a = torch.tensor([1,2,3,4])
tensor_b = torch.tensor([5,6,7,8])
```

```python
# additon
tensor_a + tensor_b
```

```python
# addition long hand
torch.add(tensor_a, tensor_b)
```

```python
# subtraction
tensor_b - tensor_a
```

```python
# subtract function
torch.sub(tensor_b, tensor_a)
```

```python
# multipilication
tensor_a * tensor_b
```

```python
torch.mul(tensor_b, tensor_a)
```

```python
# division
tensor_b / tensor_a
```

```python
torch.div(tensor_b, tensor_a)
```

```python
# remainder
tensor_b % tensor_a
```

```python
torch.remainder(tensor_b, tensor_a)
```

```python
# exponent vs power
torch.pow(tensor_a, tensor_b)
```

```python
# another way to write long hand
tensor_a.add(tensor_b)
```

```python
# assignment
tensor_a = tensor_a + tensor_b
tensor_a
```

```python
tensor_a.add_(tensor_b)
```
